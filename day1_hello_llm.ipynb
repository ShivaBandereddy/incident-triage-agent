```python
from google.colab import userdata
import os
os.environ['GOOGLE_API_KEY'] = userdata.get('GOOGLE_API_KEY')
```

```python
import os, requests
r = requests.get(
    'https://generativelanguage.googleapis.com/v1beta/models',
    headers={'x-goog-api-key': os.environ['GOOGLE_API_KEY']}, timeout=20)
print(r.status_code)
print([m['name'] for m in r.json().get('models', []) if 'flash' in m['name']])
```

```python
import os, requests
url = 'https://generativelanguage.googleapis.com/v1beta/models/gemini-3.8-flash:generateContent'
body = {'contents': [{'parts': [{'text': 'Say hello in one short line.'}]}]}
r = requests.post(url, json=body, headers={'x-goog-api-key': os.environ['GOOGLE_API_KEY']}, timeout=30)
print(r.status_code)
print(r.json()['candidates'][0]['content']['parts'][0]['text'])
```

```python
import os, requests
model = 'gemini-2.5-flash-lite'
url = f'https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent'
body = {'contents': [{'parts': [{'text': 'Say hello in one short line.'}]}]}
r = requests.post(url, json=body, headers={'x-goog-api-key': os.environ['GOOGLE_API_KEY']}, timeout=60)
print(r.status_code)
print(r.json()['candidates'][0]['content']['parts'][0]['text'])
```

```python
import os, requests
model = 'gemini-3.5-flash-lite'
url = f'https://generativelanguage.googleapis.com/v1beta/models/{model}:generateContent'
body = {'contents': [{'parts': [{'text': 'Say hello in one short line.'}]}]}
r = requests.post(url, json=body, headers={'x-goog-api-key': os.environ['GOOGLE_API_KEY']}, timeout=60)
print(r.status_code)
print(r.text[:500])
```
